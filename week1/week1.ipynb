**Make a batch W of two examples, shape (2,15)**

```python
W = np.arange(30).reshape(2, 15)
```

```python
print(W.shape)
```

or

```python
W = np.random.rand(2,15)
```

**Add a bias vector b of shape (2,) to x**

```python
b = np.arange(2)
```

```python
# @ is matrix multiplication
z = W @ x + b
```

```python
print(z)
```

**ReLU activation function: when values are negative, they become 0**

```python
a = [0 if i < 0 else i for i in z]
```

```python
print(a)
```
